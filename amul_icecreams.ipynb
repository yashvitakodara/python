```python
from selenium import webdriver
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.common.by import By
import time
from bs4 import BeautifulSoup
import requests
options = Options()
options.add_argument("--headless")
driver = webdriver.Firefox(options=options)
driver.set_window_size(1920, 1080)
```

```python
url="https://amul.com/ice-cream"
driver.get(url)
time.sleep(5)
scroll_box = """
document.querySelectorAll('div').forEach(box => {
    if (box.scrollHeight > box.clientHeight) {
        box.scrollTop = box.scrollHeight;
    }
});
"""
for _ in range(10):
    driver.execute_script(scroll_box)
    time.sleep(3)
soup = BeautifulSoup(driver.page_source, "html.parser")
cards = soup.select("div.col-lg-3.d-md-block")

for card in cards:
    n= card.select_one("span.product-name.d-lg-block").get_text(strip=True)
    q = card.select_one("span.product-size").get_text(strip=True)
    i = card.select_one("img.product-img")["src"]
    icecream.append(n)
    quantity.append(q)
    image.append(i)

```

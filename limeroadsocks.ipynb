```python
from selenium import webdriver
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.common.by import By
import time

options = Options()
options.add_argument("--headless")
driver = webdriver.Firefox(options=options)
driver.set_window_size(1920, 1080)

url = "https://www.limeroad.com/men-accessories/socks?src_id=navdeskSocks__205&src_id=navdeskSocks__305"
driver.get(url)
time.sleep(5)
scroll_box = """
document.querySelectorAll('div').forEach(box => {
    if (box.scrollHeight > box.clientHeight) {
        box.scrollTop = box.scrollHeight;
    }
});
"""
for _ in range(10):
    driver.execute_script(scroll_box)
    time.sleep(3)
i = driver.find_elements(By.CSS_SELECTOR, "img.prdI")
images = [x.get_attribute("src") for x in i]

p = driver.find_elements(By.CSS_SELECTOR, "div.dIb.vM.c3.fs14")
price = [x.text for x in p]

rating = []
products = driver.find_elements(By.CSS_SELECTOR, "div.prdC")
for product in products:
    r = product.find_elements(By.CSS_SELECTOR, "div.dIb.vM.lh12.fwB.fs11")
    if r:
        rating.append(r[0].text)
    else:
        rating.append("none")

print(len(price))
print(len(images))
print(len(rating))
driver.quit()
```

```python
import pandas as pd
df = pd.DataFrame({"image": images,"price": price,"rating":rating},index=range(1,len(price)+1))

df
```

```python
from google.colab import files
df.to_csv("limeroadsocks")
files.download("limeroadsocks")
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from google.colab import files
from urllib.parse import urljoin
import lxml
```

```python
a=[]
q=[]
abt=[]
tag=[]

url="https://quotes.toscrape.com/"
while True:
  web=requests.get(url)
  soup=BeautifulSoup(web.content,"lxml")

  q1=soup.select("span.text")
  a1=soup.select("small.author")
  abt1=soup.select("div.quote a")
  tag1=soup.select("div.tags")

  a.extend([x.text for x in a1])
  q.extend([x.text for x in q1])
  tag.extend([x.text.strip() for x in tag1])

  n=soup.select_one("li.next a")

  if n:
    next=n.get("href")
    url=urljoin(url,next)
  else:
    break
```

```python
def l(x):
  return print(len(x))
l(a)
l(q)
l(tag)
```

```python
data=pd.DataFrame({"author":a,"quote":q,"tags":tag},index=range(1,101))
print(data)
```

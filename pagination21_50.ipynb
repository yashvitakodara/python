```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from google.colab import files
from time import sleep
def f(i):
  return print(len(i))
```

```python
list1=[]
list2=[]
list3=[]
list4=[]
for x in range(1,51):
  url=f"https://books.toscrape.com/catalogue/page-{x}.html"
  web=requests.get(url)
  s=BeautifulSoup(web.content,"html.parser")
  h3=s.find_all("h3")
  l1=[x.a["title"]for x in h3]
  list1.extend(l1)
  p=s.find_all("p",class_="price_color")
  l2=[x.text for x in p]
  list2.extend(l2)
  im=s.find_all("img",class_="thumbnail")
  l3=[x["src"]for x in im]
  list3.extend(l3)
  ss1=s.find_all("p",class_="instock availability")
  l4=[x.text.strip() for x in ss1]
  list4.extend(l4)
for i in [list1,list2,list3,list4]:
  f(i)
```

```python
data=pd.DataFrame({"bookname":list1,"bookprice":list2,"coverpg":list3,"availabilty":list4},index=range(1,1001))
print(data)
# data.to_csv("last600bk")
# files.download("last600bk")
```

```python
import time
from urllib.parse import urljoin
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Starting URL
url = "https://books.toscrape.com/catalogue/page-1.html"

# Your 4 lists
list1 = []
list2 = []
list3 = []
list4 = []

# Day 2 Dynamic Loop
while url:
    print(f"Scraping: {url}")
    web = requests.get(url)
    s = BeautifulSoup(web.content, "html.parser")

    # 1. Book Titles
    h3 = s.find_all("h3")
    l1 = [x.a["title"] for x in h3]
    list1.extend(l1)

    # 2. Prices
    p = s.find_all("p", class_="price_color")
    l2 = [x.text for x in p]
    list2.extend(l2)

    # 3. Cover Images
    im = s.find_all("img", class_="thumbnail")
    l3 = [x["src"] for x in im]
    list3.extend(l3)

    # 4. Availability
    ss1 = s.find_all("p", class_="instock availability")
    l4 = [x.text.strip() for x in ss1]
    list4.extend(l4)

    # --- Day 2 'Next' Button Logic (No CSS) ---
    next_li = s.find("li", class_="next")

    if next_li and next_li.find("a"):
        relative_path = next_li.find("a")["href"]
        url = urljoin(url, relative_path)
        time.sleep(0.5)  # small pause
    else:
        print("Reached the final page!")
        url = None  # stops the while loop

# Make the DataFrame just like you did:
data = pd.DataFrame(
    {
        "bookname": list1,
        "bookprice": list2,
        "coverpg": list3,
        "availabilty": list4,
    },
    index=range(1, len(list1) + 1),
)

print(data.head())
print(f"Total rows collected: {len(data)}")
```

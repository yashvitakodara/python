```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from google.colab import files
from urllib.parse import urljoin
import lxml
```

```python
all_quotes = []
for i in range(1,11):
  url=f"https://quotes.toscrape.com/api/quotes?page={i}"
  response=requests.get(url)
  data=response.json()
  for item in data["quotes"]:

        quote_text = item["text"]
        author_name = item["author"]["name"]

        all_quotes.append({
            "quote": quote_text,
            "author": author_name
        })


df = pd.DataFrame(all_quotes)
df
```

```python
df.to_csv("quotes.api")
```

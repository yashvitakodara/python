```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import lxml
from google.colab import files
```

```python
url="https://quotes.toscrape.com/"
h= {"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:159.0) Gecko/20100101 Firefox/159.0"}
t=[]
dt=[]
lct=[]

while True:
  u=requests.get(url,headers=h)
  soup=BeautifulSoup(u.content,"html.parser")
  li=soup.select_one("li.next a")
  div=soup.select("div.quote a")
  for x in div:
    url=urljoin(url,x.get('href'))
    a=requests.Session().get(url,timeout=5)
    b=BeautifulSoup(a.content,"html.parser")
    h1=b.select("h3.author-title")
    d=b.select("span.author-born-date")
    lc=b.select("span.author-born-location")

    t.extend([x.text for x in h1])
    dt.extend([x.text for x in d])
    lct.extend([x.text for x in lc])
  if li:
    url=urljoin(url,li.get('href'))
  else:
    break
```

```python
def f(a):
  return print(len(a))
f(t)
f(dt)
f(lct)
```

```python
df = pd.DataFrame({"Author": t,"Birth_Date": dt,"Birth_Location": lct},index=range(1,101))
```

```python
df_unique = df.drop_duplicates().reset_index(drop=True)
print(df_unique)
```

```python
from bs4 import BeautifulSoup
import requests
import lxml
import pandas as pd
from google.colab import files
```

```python
url ="https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue"
h={"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/156.0.0.0 Safari/537.36 Edg/156.0.0.0"}
response=requests.get(url, headers=h)
soup = BeautifulSoup(response.content, "lxml")
table = soup.select_one("table.wikitable.sortable")

th=table.select("th")
```

```python
response
```

```python
headers=[x.text.strip() for x in th]
print(headers)
```

```python
tr=table.find_all("tr")

dataa=[]
for x in tr :
  rdta=x.find_all("td")
  dta=[x.text.strip() for x in rdta]
  dataa.append(dta)
```

```python

df=pd.DataFrame(dataa,columns=headers)

df
```

```python
df.to_csv("bigcompany.usa")
#files.download("bigcompany.usa")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from google.colab import files
```

```python
from re import L
laptop=[]
cost=[]
description=[]
reviews=[]


for x in range(1,21):
  u=f"https://webscraper.io/test-sites/e-commerce/static/computers/laptops?page={x}"
  web=requests.get(u)
  s=BeautifulSoup(web.content,"html.parser")

  a1=s.select("h4 a")
  b=s.select("h4 span")
  c=s.find_all("p",class_="description card-text")
  d=s.find_all("p",class_="review-count float-end")

  l1=[x["title"] for x in a1]
  l2=[x.text for x in b]
  l3=[x.text for x in c]
  l4=[x.text.strip() for x in d]

  laptop.extend(l1)
  cost.extend(l2)
  description.extend(l3)
  reviews.extend(l4)
```

```python
def l(y):
  return print(len(y))
l(laptop)
l(cost)
l(description)
l(reviews)
```

```python
data=pd.DataFrame({"laptop":laptop,"cost":cost,"description":description,"reviews":reviews},index=range(1,118))
data.columns = ['laptop', 'cost', 'description', 'reviews']
```

```python
data.to_csv("laptops117")
files.download("laptops117")
print(data)
```

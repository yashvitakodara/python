```python
import requests
from bs4 import BeautifulSoup
url="https://books.toscrape.com/catalogue/page-1.html"
web=requests.get(url)
print(web)
soup=BeautifulSoup(web.content,"html.parser")
```

```python
print(soup.prettify())
```

```python
print(soup.title.text)
```

```python
print(soup.div.text)
```

```python
warning=soup.find("div",class_="alert alert-warning")
print(warning.text)
```

```python
a=soup.find_all("h3")

l1=[x.a["title"] for x in a]
print(len(l1))
print(l1)
```

```python
b=soup.find_all("p",class_="price_color")
l2=[x.text for x in b]
print((l2))
print(len(l2))
```

```python
c=soup.find_all("img",class_="thumbnail")
l3=[x["src"] for x in c]
print(len(l3))
print(l3)
```

```python
d=soup.find_all("p",class_="availability")
l4=[x.text.strip() for x in d]
print(len(l4))
print(l4)
```

```python
import pandas as pandas
books=pandas.DataFrame({"Book_name":l1,"Book_price":l2,"Book_image":l3,"Availability":l4},index=range(1,21))
print(books)
```

```python
from google.colab import files

books.to_csv("booksscraped22",index=False,encoding="utf-8-sig")
files.download("booksscraped22")
```

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import pandas as pd
from google.colab import files
```

```python
url="https://books.toscrape.com"
h={"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:159.0) Gecko/20100101 Firefox/159.0"}
bn=[]
prz=[]
itag=[]
stars=[]
upc=[]
image=[]
while True:
   web=requests.get(url , headers= h)
   soup=BeautifulSoup(web.content,"html.parser")
   next=soup.select_one("li.next a")
   linc=soup.select("h3 a")
   for x in linc:
     burl=urljoin(url,x.get('href'))
     web2=requests.get(burl, headers=h)
     soup2=BeautifulSoup(web2.content,"html.parser")

     h1=soup2.select("h1")
     p=soup2.select("div.row p.price_color")
     i_tags = soup2.select("div.row p.star-rating i")
     u=soup2.select_one('td')
     img=soup2.select("div.row img")


     bn.extend([x.text for x in h1])
     prz.extend([x.text for x in p])
     stars.extend([len(i_tags)])
     upc.extend([u.text.strip()] if u else "")
     image.extend([x.get('src') for x in img])

   if next:
    url = urljoin(url, next.get("href"))
   else:
    break
```

```python
print(len(bn))
print(len(prz))
print(len(stars))
print(len(upc))
print(len(image))
```

```python
df = pd.DataFrame({"Book Name": bn, "Price": prz, "Stars": stars, "UPC": upc,"Imagelinks":image},index=range(1,1001))
df.to_csv("datacrawlled")
```

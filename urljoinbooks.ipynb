```python
import time
from urllib.parse import urljoin
import pandas as pd
import requests
from bs4 import BeautifulSoup

url ="https://books.toscrape.com/catalogue/page-1.html"

list1 = []
list2 = []
list3 = []
# list4 = []
h= {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0.0.0 Safari/537.36"}
while True:
    web=requests.get(url, headers=h)
    soup=BeautifulSoup(web.content,"html.parser")
    pg_no=soup.select("li.current")

    n=soup.select("h3 a")
    prize=soup.select("p.price_color")
    img=soup.select("img.thumbnail")

    list1.extend([x.text for x in n])
    list2.extend([x.test for x in prize])
    list3.extend([x.get("src") for x in img])

    npg=soup.select_one("li.next a")
    if npg:
      nurl=npg.get("href")
      url=urljoin(url,nurl)

    else:
      print("the end")
      break
```
